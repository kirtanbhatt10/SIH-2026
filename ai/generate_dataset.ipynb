```python
from pathlib import Path
from ml.dataset_v2.validate_dataset import validate_dataset

DATASET_DIR = Path("training_data_v2_test")

checks = validate_dataset(DATASET_DIR)

print("=" * 70)
print("DATASET V2 VALIDATION RESULTS")
print("=" * 70)

passed = 0
failed = 0

for check in checks:
    # Show the Check object's fields
    print(check)
    
    # Try common attribute names
    status = getattr(check, "passed", None)
    name = getattr(check, "name", "Unknown check")
    message = getattr(check, "message", "")
    
    if status is True:
        passed += 1
        print(f"✅ PASS: {name} {message}")
    elif status is False:
        failed += 1
        print(f"❌ FAIL: {name} {message}")
    else:
        print("   ↳ Check fields:", vars(check))

print("\n" + "=" * 70)
print(f"PASSED: {passed}")
print(f"FAILED: {failed}")
print("=" * 70)
```

```python
from pathlib import Path
import json
import numpy as np

DATASET_DIR = Path("ml/training_data_v2")

print("=" * 70)
print("DATASET V2 METADATA / REPRODUCIBILITY CHECK")
print("=" * 70)

# Load metadata
with open(DATASET_DIR / "dataset_info.json", "r", encoding="utf-8") as f:
    info = json.load(f)

print("Dataset version :", info.get("dataset_version"))
print("Generator       :", info.get("generator_version"))
print("Seed            :", info.get("seed"))
print("Samples         :", info.get("n_samples"))
print("Features        :", info.get("n_features"))
print("Sample rate     :", info.get("sample_rate"))
print("Chunk size      :", info.get("chunk_size"))

# Load arrays
X = np.load(DATASET_DIR / "features.npy")
y = np.load(DATASET_DIR / "labels.npy")

print("\nActual arrays:")
print("features shape  :", X.shape)
print("labels shape    :", y.shape)
print("dtype           :", X.dtype)

# Basic deterministic fingerprint
import hashlib

x_hash = hashlib.sha256(X.tobytes()).hexdigest()
y_hash = hashlib.sha256(y.tobytes()).hexdigest()

print("\nDataset fingerprints:")
print("features SHA256 :", x_hash)
print("labels SHA256   :", y_hash)

print("\nSTATUS: ✅ METADATA + FINGERPRINT CHECK COMPLETE")
```
